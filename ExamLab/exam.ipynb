```python
def deco(func):
    def wrapper(*args,**kwargs):
        print("Start")
        res=func(*args,**kwargs)
        print(res)
        print("End")
        
    return wrapper
@deco
def add(a,b):
    return a+b

@deco
def subtract(a,b):
    return b-a
add(2,3)
subtract(2,3)
```

```python
import logging
logging.basicConfig(level=logging.WARN,format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',filename="a.log",filemode="a")
logger=logging.getLogger(__name__)
class Bankaccount:
    def __init__(self,balance=0):
        self.__balance=balance
    def __str__(self):
        return f"{self.__balance}"
    def deposit(self, amt):
        if amt <= 0:
            logger.warning("Deposit amount must be positive")
        else:
            self.__balance += amt
            logger.warning(f"Deposited {amt}. New balance: {self.__balance}")
    def withdraw(self, amt):
        if amt > self.__balance:
            logger.warning("Insufficient balance")
        else:
            self.__balance -= amt
            logger.warning(f"Withdrew {amt}. New balance: {self.__balance}")

B=Bankaccount(1000)
B.deposit(100)
print(B)
B.withdraw(1000)
print(B)

    
```

```python
class Animal:
    def __init__(self,name):
        self.name=name
    def __str__(self):
        return f"The name of the dog is {self.name}"
class Dog(Animal):
    def __init__(self,name,breed):
        super().__init__(name)
        self.breed=breed
    def __str__(self):
        return f"{super().__str__()}\nThe breed is {self.breed}"

d=Dog("hh","heroic")
print(d)
```

```python
import sqlite3

# Connect to database (creates file if not exists)
conn = sqlite3.connect("example.db")
cursor = conn.cursor()

# 1. Create table
cursor.execute("""
CREATE TABLE IF NOT EXISTS students(
    id INTEGER PRIMARY KEY,
    name TEXT
)
""")

# 2. Insert rows (OR IGNORE avoids duplicate primary keys)
cursor.execute("INSERT OR IGNORE INTO students VALUES(1, 'Ayush')")
cursor.execute("INSERT OR IGNORE INTO students VALUES(3, 'Ayush')")
conn.commit()

cursor.execute("INSERT OR IGNORE INTO students VALUES(2, 'Ayush')")
conn.commit()

# 3. Update row
cursor.execute("UPDATE students SET name='Ram' WHERE id=1")

# 4. Delete row
cursor.execute("DELETE FROM students WHERE id=1")
conn.commit()   # 🔥 important to save changes

# 5. Select rows
cursor.execute("SELECT id, name FROM students")
rows = cursor.fetchall()   # 🔥 important to fetch results

# 6. Print results
print(rows)

# Close connection (good practice)
conn.close()
```

```python
import pandas as pd

df_data = pd.DataFrame({
    'Name': ['Ram','Sita','Ram','Sita','Ram','Sita','Ram','Sita'],
    'Test': ['Test1','Test1','Test1','Test1','Test2','Test2','Test2','Test2'],
    'Subject': ['Math','Science','Math','Science','Math','Science','Math','Science'],
    'Marks': [80, 85, 75, 90, 88, 92, 70, 95]
})

print(df_data)
pivot = pd.pivot_table(
    df_data,
    index='Name',              # rows
    columns=['Test','Subject'],# columns
    values='Marks',            # what to aggregate
    aggfunc='mean',            # average marks
    fill_value=0
)

print(pivot)
```
