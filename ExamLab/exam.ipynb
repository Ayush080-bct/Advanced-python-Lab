```python
def deco(func):
    def wrapper(*args,**kwargs):
        print("Start")
        res=func(*args,**kwargs)
        print(res)
        print("End")
        
    return wrapper
@deco
def add(a,b):
    return a+b

@deco
def subtract(a,b):
    return b-a
add(2,3)
subtract(2,3)
```

```python
import logging
logging.basicConfig(level=logging.WARN,format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',filename="a.log",filemode="a")
logger=logging.getLogger(__name__)
class Bankaccount:
    def __init__(self,balance=0):
        self.__balance=balance
    def __str__(self):
        return f"{self.__balance}"
    def deposit(self, amt):
        if amt <= 0:
            logger.warning("Deposit amount must be positive")
        else:
            self.__balance += amt
            logger.warning(f"Deposited {amt}. New balance: {self.__balance}")
    def withdraw(self, amt):
        if amt > self.__balance:
            logger.warning("Insufficient balance")
        else:
            self.__balance -= amt
            logger.warning(f"Withdrew {amt}. New balance: {self.__balance}")

B=Bankaccount(1000)
B.deposit(100)
print(B)
B.withdraw(1000)
print(B)

    
```

```python
class Animal:
    def __init__(self,name):
        self.name=name
    def __str__(self):
        return f"The name of the dog is {self.name}"
class Dog(Animal):
    def __init__(self,name,breed):
        super().__init__(name)
        self.breed=breed
    def __str__(self):
        return f"{super().__str__()}\nThe breed is {self.breed}"

d=Dog("hh","heroic")
print(d)
```

```python
import sqlite3

conn = sqlite3.connect("example.db")
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS students(
    id INTEGER PRIMARY KEY,
    name TEXT
)
""")

cursor.execute("""
INSERT OR IGNORE INTO students VALUES(1, "Ayush")
""")

conn.commit()   # 🔥 important

cursor.execute("SELECT id, name FROM students")
rows = cursor.fetchall()   # 🔥 important

print(rows)
```
