```python
# Create a Car class with attributes make , model , and year . Add a method to display car
# details.
class Car:
    def __init__(self,make,model,year):
        self.make=make
        self.model=model
        self.year=year
    def __str__(self):
        return f"Manufacture by : {self.make.upper()}\nModel : {self.model}\nYear Made : {self.year}"
C1=Car("Tesla","Model A",2023)
print(C1)
```

```python
# Extend the Car class to ElectricCar with an additional attribute battery_capacity . Add a
# method to display battery info.
class ElectricCar(Car):
    def __init__(self, make, model, year, battery_capacity):
        super().__init__(make, model, year)
        self.battery_capacity = battery_capacity
    
    def display_battery_info(self):  # A dedicated method as requested
        return f"Battery Capacity: {self.battery_capacity} kWh"
    
    def __str__(self):
        return f"{super().__str__()}\n{self.display_battery_info()}"

EC1 = ElectricCar("Tesla", "Model 3", 2024, 79)
print(EC1)  
EC1.display_battery_info()
```

```python
class Rectangle:
    def __init__(self,length,breadth):
        self.length=length
        self.breadth=breadth
    def __str__(self):
        return f"Rectangle({self.length}cm,{self.breadth}cm)"
    def CalArea(self):
        return f"Area of the {str(self)} ={self.length*self.breadth} sq.cm"
    def CalPreme(self):
        return f"Premeter of the {str(self)}= {2*(self.length+self.breadth)} cm"
R=Rectangle(5,4)
print(R)
print(R.CalArea())
print(R.CalPreme())
```

```python
# Create a BankAccount class. Add methods for deposit, withdraw, and checking balance. Ensure
# the balance cannot be accessed directly.
import logging
logging.basicConfig(level=logging.DEBUG,
                    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s')
logger=logging.getLogger(__name__)
class BankAccount:
    def __init__(self,balance=0):
        self.__balance=balance
    def deposit(self, amt):
        if amt <= 0:
            logger.warning("Deposit amount must be positive")
        else:
            self.__balance += amt
            logger.info(f"Deposited {amt}. New balance: {self.__balance}")
    def withdraw(self, amt):
        if amt > self.__balance:
            logger.warning("Insufficient balance")
        else:
            self.__balance -= amt
            logger.info(f"Withdrew {amt}. New balance: {self.__balance}")
    @property
    def balance(self):
        return self.__balance
B=BankAccount(1000)
B.deposit(1000)
print(f"The current balance in account is {B.balance}")
            
print(B.withdraw(1200))
print(f"The current balance in account is {B.balance}")
            
        
```

```python
# Create a Student class with attributes name , roll_number , and marks . Add methods to
# display student info and calculate average marks.
import numpy as np
class Student:
    def __init__(self,name,roll_number,marks:np.array):
        self.name=name
        self.roll_number=roll_number
        self.marks=marks
    def __str__(self):
        return f"{'-'*5}StudentInfo{'-'*5}\nName: {self.name}\nRoll_number: {self.roll_number}\nmarks: {self.marks}"
    def display(self):
        return f"Average marks={np.mean(self.marks)}"
S=Student("John cena",16,np.array([80,70,80,90,90]))
print(S)
print(S.display())
```

```python
#     Create a Shape class and two subclasses Triangle and Square . Implement area calculation
# for both using inheritance.
class Shape:
    def area(self):
        pass
class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def area(self):
        return 0.5 * self.base * self.height
class Square(Shape):
    def __init__(self, side):
        self.side = side
    def area(self):
        return self.side * self.side
tri=Triangle(4, 6)
print("Triangle area:", tri.area())

sq=Square(4)
print("Square area:", sq.area())
```

```python
# Implement a Teacher class that inherits from Person . Override the greet method to include
# subject taught.
class Person:
    def __init__(self,name):
        self.name=name
    def greet(self):
        return f"Hello {self.name}"
class Teacher(Person):
    def __init__(self,name,subject):
        super().__init__(name)
        self.subject=subject
    def greet(self):
        return f"{super().greet()} and you are teaching {self.subject}!"

T = Teacher("Aayush", "Math")
print(T.greet())

```

```python
# create a LibraryBook class with attributes title , author , available . Add methods to
# borrow and return the book. Use encapsulation for available .
class LibraryBook:
    def __init__(self, title, author, available=1):
        self.title = title
        self.author = author
        self.__available = available 
    @property
    def available(self):
        return self.__available > 0
    def borrowBook(self):
        if self.__available > 0:
            self.__available -= 1
            return f"Here is the book you requested:\nTitle: {self.title}\nAuthor: {self.author}"
        else:
            raise ValueError("Sorry! The book is not available.")
    def returnBook(self):
        self.__available += 1
        return f"Thank you for returning '{self.title}'. Copies available now: {self.__available}"
B=LibraryBook("Harry Putteeerrr","John cena",2)
print(B.borrowBook())
print(B.borrowBook())
print(B.available)
```

```python
# Implement a CompanyEmployee class. Include name , salary , and position . Add methods
# to display info and give a raise. Demonstrate encapsulation for salary.
class CompanyEmployee:
    def __init__(self,name,salary,position="HR"):
        self.name=name
        self.__salary=salary
        self.position=position
    @property
    def salary(self):
        return self.__salary
    def __str__(self):
        return f"{'-'*5}Employee{'-'*5}\nName = {self.name}\nSalary = {self.__salary}\nPosition = {self.position}"
    @salary.setter
    def salary(self, amt):
        if amt < 0:
            raise ValueError("Raise amount must be positive.")
        self.__salary += amt

C=CompanyEmployee("John",20000,"HR")
print(C)
C.salary=1000
print(C)
```
